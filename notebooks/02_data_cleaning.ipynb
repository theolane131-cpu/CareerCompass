# Data Cleaning

This notebook loads the original raw dataset and performs the initial cleaning steps. The cleaned dataset will be saved separately so that the original raw data remains unchanged throughout the project.

```python
# Import the Pandas library for data manipulation and analysis
import pandas as pd

df = pd.read_csv("../data/raw/ds_salaries.csv")

df.head()
```

```python
# Remove the unnecessary index column created during CSV export
df = df.drop(columns=["Unnamed: 0"])

df.head()
```

```python
# Print the remaining column names to verify the dataset structure
print(df.columns)
```

```python
# Save the cleaned dataset to the processed data folder for future use
df.to_csv("../data/processed/processed_salary_data.csv", index=False)
```
