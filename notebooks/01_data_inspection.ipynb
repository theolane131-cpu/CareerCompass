# Initial Data Inspection

This notebook performs an initial inspection of the employee salary dataset. The purpose is to understand the dataset's structure, identify potential data quality issues, and gather summary information before data cleaning and machine learning model development.

## Step 1: Import Required Libraries

```python
# Import the Pandas library for data loading and analysis
import pandas as pd

# Load the original salary dataset from the raw data folder
df = pd.read_csv("../data/raw/ds_salaries.csv")

# Display the first five rows to verify the dataset loaded correctly
df.head()
```

## Step 2: Determine the Dataset Size

```python
# Display the number of rows and columns in the dataset
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## Step 3: Display the Dataset Columns

```python
# Display the names of all columns in the dataset
print(df.columns.tolist())
```

## Step 4: Inspect Data Types and Missing Values

```python
# Display information about each column including data type and non-null values
df.info()
```

## Step 5: Check for Missing Values

```python
# Count the number of missing values in each column
df.isnull().sum().sort_values(ascending=False)
```

## Step 6: Check for Duplicate Records

```python
# Count duplicate rows in the dataset
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

## Step 7: Generate Summary Statistics

```python
# Generate summary statistics for both numerical and categorical features
df.describe(include="all").T
```

## Step 8: Examine the Target Variable

```python
# Display descriptive statistics for the salary prediction target variable
df["salary_in_usd"].describe()
```

## Step 9: Count Unique Values

```python
# Display the number of unique values contained in each feature
for column in df.columns:
    print(f"{column}: {df[column].nunique()} unique values")
```
