# Feature Engineering

This notebook prepares the cleaned employee salary dataset for machine learning model development. The preprocessing steps include loading the cleaned dataset, selecting predictor variables, encoding categorical features, splitting the data into training and testing sets, and saving the processed datasets for future model training and evaluation.

```python
# Import libraries used for feature engineering
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
# Load the cleaned dataset from Deliverable 1
df = pd.read_csv("../data/processed/processed_salary_data.csv")

# Display the first five rows
df.head()
```

```python
# Display information about the dataset
df.info()
```

```python
# Separate predictor variables from the target variable
# Remove salary and salary_currency to prevent data leakage

X = df.drop(columns=[
    "salary_in_usd",
    "salary",
    "salary_currency"
])

# Target variable
y = df["salary_in_usd"]

# Display predictor variables
print("Predictor Variables:")
print(X.columns.tolist())

print("\nTarget Variable:")
print(y.name)
```

## Encoding Categorical Variables

Machine learning algorithms require numerical input features. Since several predictor variables in this dataset are categorical, One-Hot Encoding is used to convert categorical values into binary indicator variables. This approach prevents the model from assuming an artificial numerical ordering between categories while preserving the information contained in each feature.

```python
# Apply One-Hot Encoding to all categorical predictor variables

X_encoded = pd.get_dummies(
    X,
    drop_first=True
)

# Display the first five rows
X_encoded.head()
```

```python
# Display the new dimensions after encoding

print("Original Features:", X.shape)

print("Encoded Features:", X_encoded.shape)
```

```python
# Display the encoded feature names with an index number
feature_df = pd.DataFrame({
    "Feature Number": range(1, len(X_encoded.columns) + 1),
    "Encoded Feature Name": X_encoded.columns
})

feature_df
```

```python
print(f"Total Encoded Features: {len(X_encoded.columns)}")
```

```python
# Display the dimensions before and after one-hot encoding
print("Feature Engineering Summary")
print("-" * 40)

print(f"Original Feature Matrix : {X.shape}")
print(f"Encoded Feature Matrix  : {X_encoded.shape}")
print(f"Total Encoded Features  : {len(X_encoded.columns)}")
```

## Train-Test Split

To evaluate model performance fairly, the dataset is divided into training and testing sets. The training data is used to build the machine learning models, while the testing data is reserved for evaluating model performance on unseen data. An 80/20 split is used, which is a common practice for supervised machine learning tasks.

```python
# Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Display the size of each dataset

print("Training Features:", X_train.shape)
print("Testing Features :", X_test.shape)

print()

print("Training Labels :", y_train.shape)
print("Testing Labels  :", y_test.shape)
```

## Save Processed Modeling Data

The processed training and testing datasets are saved for reuse during model training and evaluation. Saving these datasets improves reproducibility and keeps the machine learning pipeline organized.

```python
# Save processed datasets for future notebooks

X_train.to_csv("../data/processed/X_train.csv", index=False)
X_test.to_csv("../data/processed/X_test.csv", index=False)

y_train.to_csv("../data/processed/y_train.csv", index=False)
y_test.to_csv("../data/processed/y_test.csv", index=False)

print("Processed datasets saved successfully.")
```
